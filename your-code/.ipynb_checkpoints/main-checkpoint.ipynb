#### Train, Test, Split

1. Import `train_test_split` and split your data accordingly. Choose an appropriate test size.

```python
from sklearn.model_selection import train_test_split
SEED = 42

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=SEED)
```

3. There is another metric called Root mean squared error, Which is the square root of the MSE. Calculate the RMSE.

```python
from sklearn.metrics import mean_squared_error, r2_score

print(
    f"Train  R² = {r2_score(y_train, y_train_pred):.4f},"
    f"  MSE = {mean_squared_error(y_train, y_train_pred):.2f}"
)
print(
    f"Test   R² = {r2_score(y_test, y_pred):.4f},"
    f"  MSE = {mean_squared_error(y_test, y_pred):.2f}"
)
```

### Evaluation — $R^2$ and MSE

$$R^2 = 1 - \frac{\sum(y - \hat{y})^2}{\sum(y - \bar{y})^2}$$

* $R^2 = 1$ → perfect fit.
* $R^2 = 0$ → no better than predicting the mean.

$R^2$ tells us the **proportion of variance** in the target that our model explains. An $R^2$ of 0.7 means the model captures 70% of the variability in the data.

$$\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

MSE is the **average squared error** — it tells us how far off our predictions are, in squared units of the target. Lower is better. Taking the square root gives RMSE, which is in the same units as $y$ and easier to interpret.
